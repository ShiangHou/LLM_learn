```python
#
# @lc app=leetcode.cn id=394 lang=python
#
# [394] 字符串解码
#
# 输入：s = "3[a]2[bc]"
# 输出："aaabcbc"

# @lc code=start
def decodeString(s):
    #栈，如果是数字，字母，[,统统压入如果是]，就开始解码，解码到纯字母停止，然后放进去就行
    stack = []
    for i in s:
        if i.isdigit() or i.isalpha() or i == "[":
            stack.append(i)
            continue 
        elif i == "]":
            temp_alpha = ""
            while stack[-1] != "[":
                temp = stack.pop()
                temp_alpha += temp
                
            temp_alpha = temp_alpha[::-1]
            stack.pop()
            if not stack or stack[-1].isalpha():#前面没数字，或者就是字母，直接加进去
                stack.append(temp_alpha)
            else:
                temp_digit = ""
                while stack and stack[-1].isdigit():
                    temp_num = stack.pop()
                    temp_digit += temp_num 
                nums = int(temp_digit[::-1])
                #乘好加进去
                
                stack.append(nums*temp_alpha)
    return  "".join(stack)

    
s =  "3[a]2[bc]"
decodeString(s)



        
# @lc code=end

```
